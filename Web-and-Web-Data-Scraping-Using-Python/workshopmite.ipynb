```python
class Person:
    def __init__(self, name):
        self.name = name

    def greet(self):
        print("Hello, My Name is:", self.name)

    def __str__(self):
        return 'Person: ' +self.name

p = Person("Ravi")
p.greet()
print("\n")
print(p)
```

```python
class Animal:
  def speak(self):
    print("Animal Speaks")


class Dog(Animal):
  def speak(self):
    print("Dog Barks")
  pass
d=Dog()
d.speak()
```

Polymorphism

```python
class bird:
  def fly(self):
    print("Bird Flies")

class Airplane:
  def fly(self):
    print("Airplane Flies")

class Sky:
  def fly(self):
    print("Sky Flies")

def make_it_fly(obj):
  obj.fly()

make_it_fly(bird())
make_it_fly(Airplane())
make_it_fly(Sky())
```

Composition

```python
class Engine:
  def start(self):
    print("Engine Started")

class car:
  def __init__(self):
    self.engine=Engine()

  def drive(self):
    self.engine.start()
    print("car Moving")

c=car()
c.drive()
```

```python
class InsufficientBalancError(Exception):
  pass

class Bank_Account:
  def __init__(self,balance):
    self.balance=balance

  def withdraw(self,amount):
    if amount>self.balance:
      raise InsufficientBalancError("Not Enough Balance")
    self.balance-=amount


try:
  acc=Bank_Account(1000)
  acc.withdraw(1500)
  print("Amount Withdrawn")
except InsufficientBalancError as e:
  print("Error: ",e)
```
