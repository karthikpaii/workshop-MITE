Calss and Object

```python
class Person:
    def __init__(self, name):
        self.name = name

    def greet(self):
        print("Hello, My Name is:", self.name)

    def __str__(self):
        return 'Person: ' +self.name

p = Person("Ravi")
p.greet()
print("\n")
print(p)
```

Inheritance

```python
class Animal:
  def speak(self):
    print("Animal Speaks")


class Dog(Animal):
  def speak(self):
    print("Dog Barks")
  pass
d=Dog()
d.speak()
```

Polymorphism

```python
class bird:
  def fly(self):
    print("Bird Flies")

class Airplane:
  def fly(self):
    print("Airplane Flies")

def make_it_fly(obj):
  obj.fly()

make_it_fly(bird())
make_it_fly(Airplane())

```
